```python
import xml.etree.ElementTree as ET

report = CODE_DIR / "tests" / "_report.xml"
pt("tests", "-q", f"--junitxml={report}")

rows = []
for tc in ET.parse(report).getroot().iter("testcase"):
    if tc.find("failure") is not None or tc.find("error") is not None:
        status = "FAIL"
    elif tc.find("skipped") is not None:
        status = "skip"
    else:
        status = "pass"
    rows.append((status, float(tc.get("time", 0)), tc.get("classname", "").split(".")[-1], tc.get("name")))

print()
for status, secs, mod, name in sorted(rows, key=lambda r: (r[0] != "FAIL", r[2], r[3])):
    print(f"{status:5s} {secs:6.2f}s  {mod}::{name}")
n_fail = sum(r[0] == "FAIL" for r in rows)
print(f"\n{len(rows)} tests: {sum(r[0]=='pass' for r in rows)} passed, {n_fail} failed, {sum(r[0]=='skip' for r in rows)} skipped")
```

```python
import subprocess, sys
from pathlib import Path

CODE_DIR = Path.cwd()
assert (CODE_DIR / "tests").is_dir(), f"tests/ not found in {CODE_DIR}; start the notebook from the code folder"

def pt(*args):
    """Run pytest with the kernel's own Python and stream its output into this cell."""
    proc = subprocess.Popen(
        [sys.executable, "-m", "pytest", *args],
        cwd=CODE_DIR, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True,
    )
    for line in proc.stdout:
        print(line, end="")
    return proc.wait()

# -v   one line per test
# -rsx list WHY tests were skipped (s) or xfailed (x) at the end
# --tb=short  compact tracebacks
exit_code = pt("tests", "-v", "-rsx", "--tb=short", "--color=yes")
print("\nALL TESTS PASSED" if exit_code == 0 else f"\nFAILURES (pytest exit code {exit_code})")
```
